```python
top_10_locations = analysis_df['start'].value_counts().head(10)
top_10_locations.plot(kind='bar', title='Top 10 Start Locations')
```

```python
peak_hr_analysis = analysis_df['is_peak_hr'].value_counts().plot(kind='bar', title='Peak Hour Analysis')
```

```python
weekend_analysis = analysis_df['is_weekend'].value_counts().plot(kind='bar', title='Weekend Analysis')
```

```python
start_with_avg_duration = analysis_df.groupby('start')['duration_minutes'].mean().head(15).sort_values(ascending=False)
start_with_avg_duration.plot(kind='bar', title='Avg Duration(min) by Start Location')
```

```python
hourly_demand = analysis_df.groupby('hour_of_day')['duration_minutes'].count().sort_values(ascending=False)
hourly_demand.plot(kind='bar', title='Hourly Demand')
```

```python
weekly_demand = analysis_df.groupby('day_of_week_name')['duration_minutes'].count().sort_values(ascending=False)
weekly_demand.plot(kind='bar', title='Weekly Demand')
```

```python
stt_miles = analysis_df.groupby('start')['miles'].mean().head(15).sort_values(ascending=False)
stt_miles.plot(kind='bar', title='Avg Miles by Start Location')
```

```python
#monthly trend
daily_count = analysis_df.groupby(['month', 'day']).size().reset_index(name = "total_rides")
daily_count = daily_count.groupby('month')['total_rides'].sum()
daily_count.plot(kind = 'bar' , ax=ax[1,1] , title = 'Monthly Pattern')
fig
```

```python
miles_dur = sns.scatterplot(data = analysis_df, x = 'miles' , y = 'duration_minutes' , hue = "category" , ax=ax[2,0] )
ax[2,0].set_title("distance vs duration")
fig
```

```python
pivot = pd.crosstab(
    analysis_df['day_of_week_name'],
    analysis_df['hour_of_day'] 
)
days = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]
pivot = pivot.reindex(days)
target_ax=ax[2,1]
target_ax.imshow(pivot , aspect = 'auto')
target_ax.set_xticks(range(len(pivot.columns)))
target_ax.set_xticklabels(pivot.columns)

target_ax.set_yticks(range(len(pivot.index)))
target_ax.set_yticklabels(pivot.index)

target_ax.set_xlabel("Hour of Day")
target_ax.set_ylabel("Day of Week")
target_ax.set_title("Ride Demand by Day and Hour")
fig
pivot
```
