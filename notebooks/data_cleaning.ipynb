```python
df['start'] = pd.to_datetime(df['START_DATE'] , format= 'mixed' , errors='coerce')
df['end'] = pd.to_datetime(df['END_DATE'], format='mixed', errors='coerce')
# new_df = df[['start', 'end']] #changes to new_df will lead chnages to acual df as well
new_df = df[['start', 'end']].copy() #changes to new_df will not lead chnages to acual df as well
new_df['start_date'] = df['start'].dt.date
new_df['start_time'] = df['start'].dt.time
new_df['end_date'] = df['end'].dt.date
new_df['end_time'] = df['end'].dt.time
new_df
```

```python
new_df['duration'] = new_df['end'] - new_df['start']
new_df 
# ACTUAL DATETIME HI SUBTRACT KR SKTE HAI
# CONVERTED NAI KR SKTE HAI
```

```python
new_df['miles'] = df['MILES']
new_df['category'] = df['CATEGORY']
new_df['purpose'] = df['PURPOSE']
new_df['start'] = df['START']
new_df['end'] = df['STOP']
new_df
```

```python
new_df['duration_minutes'] = new_df['duration'].dt.total_seconds() / 60
new_df
```

```python
new_df.drop(1155)
```

```python
new_df['month'] = df['start'].dt.month
new_df['day'] = df['start'].dt.day
new_df
```

```python
new_df['hour_of_day'] = df['start'].dt.hour
new_df['day_of_week'] = df['start'].dt.dayofweek
new_df['day_of_week_name'] = df['start'].dt.day_name()
new_df
```

```python
clean_df = new_df.copy()
clean_df.drop('start_date' , axis=1, inplace=True)
clean_df.drop('end_date' , axis=1, inplace=True)
clean_df.drop('start_time' , axis=1, inplace=True)
clean_df.drop('end_time' , axis=1, inplace=True)
clean_df.drop(1155 , axis=0, inplace=True)
clean_df.info()
```
